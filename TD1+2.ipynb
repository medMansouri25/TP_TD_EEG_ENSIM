## 1. Contexte général

La conduite automobile est une activité complexe qui mobilise plusieurs ressources cognitives : perception visuelle, attention soutenue, prise de décision, anticipation, contrôle moteur et gestion des imprévus.

Le papier CL-Drive s'intéresse donc à une question centrale :

> Peut-on estimer automatiquement la charge cognitive d'un conducteur à partir de signaux physiologiques et comportementaux mesurés pendant la conduite ?

#### Q1. Pourquoi est-il intéressant d'estimer la charge cognitive d'un conducteur ?

**Réponse :**
Un grand nombre d'accidents sont causés par des conducteurs distraits ou somnolents.
Leur distraction peut venir d'une charge cognitive trop élevée due à des tâches secondaires. Si la charge cognitive dépasse un seuil, les performances du conducteur se dégradent et cela augmente le risque d'accident. Ainsi, il serait utile d'intégrer dans les véhicules intelligents un système capable de mesurer la charge cognitive et d'alerter le conducteur si celle-ci est trop élevée.

## 2. Définition : charge cognitive

#### Q2. Comment définir la charge cognitive ?

**Réponse :**

*(Section 1 — Introduction, paragraphe 1)*

La charge cognitive c'est la quantité d'informations que notre mémoire de travail peut traiter à un instant donné, autrement dit la quantité de ressources cognitives nécessaires pour accomplir une tâche. Il y a deux types :
- La **charge intrinsèque** : liée à la complexité de la tâche elle-même.
- La **charge extrinsèque** : liée à des facteurs environnementaux qui n'ont rien à voir avec la tâche.


#### Q3. Dans le contexte de la conduite, de quels facteurs la charge cognitive peut-elle dépendre ?

**Réponse :**

*(Section 1 — Introduction, paragraphe 1 pour le dernier point ; Section 2.2 — Cognitive Load in Driving, paragraphe 1 pour les tâches secondaires)*

- L'engagement dans des tâches secondaires comme une conversation verbale, des jeux de mots, ou l'interaction avec un iPod.
- La tâche de conduite elle-même : le document souligne que la plupart des travaux oublient que conduire peut en soi être une forte source de charge cognitive élevée.

Dans le papier, la charge cognitive est évaluée à l'aide d'une échelle subjective appelée **PAAS**, allant de 1 à 9.

#### Q5. Pourquoi parle-t-on ici d'une vérité terrain subjective ?

**Réponse :**

*(Section 3.6 — Cognitive Load Self-Assessment, paragraphe 1)*

C'est subjectif parce que c'est chaque participant qui dit lui-même ce qu'il ressent, sur le moment. Un signal sonore se déclenche toutes les 10 secondes pour rappeler au participant de donner son score à voix haute, et un membre de l'équipe le note. Ce n'est pas un capteur qui mesure quelque chose d'objectif. Ainsi, deux personnes dans le même scénario peuvent donner des scores complètement différents selon comment elles vivent la situation.

## 5. Modalités mesurées

Le dataset CL-Drive est **multimodal**. Il contient plusieurs types de signaux enregistrés pendant la même tâche de conduite.

#### Q7. Compléter le tableau suivant.

| Modalité | Signification | Signal mesuré | Rôle possible |
|---|---|---|---|
| EEG | Électroencéphalographie | Activité électrique du cerveau, 4 canaux AF7/AF8/TP9/TP10, 256 Hz | Reflète l'activité neuronale liée à la charge cognitive, notamment dans les régions fronto-pariétales |
| ECG | Électrocardiographie | Activité électrique du cœur, signaux LL-RA/LA-RA/Vx-RA, 512 Hz | Le système nerveux autonome qui contrôle la variabilité du rythme cardiaque est impacté par la charge cognitive |
| EDA | Activité électrodermale | Conductance cutanée, 2 électrodes abdomen gauche, 128 Hz | Le système nerveux sympathique qui contrôle la réponse cutanée est impacté par la charge cognitive |
| Gaze | Suivi oculaire | Saccades, fixations, diamètre pupillaire, clignements, 50 Hz | Corrélations significatives entre changements de taille de pupille, taux de clignement, saccades, fixations et charge cognitive |

*Sources : Section 3.1 — Sensors paragraphes EEG, ECG, EDA, Gaze ; Section 2.3 — BCI paragraphes 2, 3 et 4*


Dans le projet, la modalité principale sera l’**EEG**. Les autres modalités serviront éventuellement d’extensions pour discuter de la fusion multimodale.

#### Q8. Pourquoi peut-il être intéressant de combiner EEG, ECG, EDA et Gaze pour estimer la charge cognitive ?

**Réponse :**
*(Section 2.3 — BCI, paragraphes 2, 3 et 4 ; Section 5 — Benchmarking Results)*

Chaque modalité capture un aspect différent de la réponse physiologique. Cette approche multimodale permet d'améliorer la précision de l'estimation de la charge cognitive. Le système nerveux sympathique (EDA) et le système nerveux autonome (ECG) sont tous les deux impactés par la charge cognitive. Les changements de taille de pupille, de taux de clignement, de saccades et fixations (Gaze) sont aussi corrélés à la charge cognitive. Les résultats de benchmarking confirment que les configurations avec les 4 modalités obtiennent systématiquement les meilleures performances.

## 7. Protocole expérimental

#### Q11. Quelle est la différence entre un **protocole expérimental** et un **dataset exploitable en machine learning** ?

**Réponse :**

*(Section 3 — Experiment Setup and Data Collection ; Section 4 — Data Processing)*

Le **protocole expérimental** décrit comment les données ont été collectées : qui sont les participants, quels capteurs, quels scénarios, dans quel ordre, comment les labels sont recueillis.

Le **dataset exploitable en machine learning** est le résultat après traitement : des paires (segment de signal, label) prêtes à entraîner un modèle. Cela implique le prétraitement, la segmentation, la normalisation et l'extraction de features.

#### Q12. Donner une description détaillée du protocole expérimental utilisé.

**Réponse :**

1. **Recrutement**
    * Les données ont été collectées auprès de 23 participants, dont 17 femmes et 6 hommes. Au final, 21 participants ont été retenus.
    * Chaque participant a signé un consentement écrit et l'étude a été approuvée par le comité d'éthique de Queen's University (GREB).

    (Section 3.5 — Participants, paragraphe 1)
2. **Installation des capteurs** :
    * Le casque EEG Muse S, les capteurs Shimmer ECG et EDA, et les lunettes Tobii ont été placés sur le participant, puis connectés via Bluetooth à une station d'acquisition.

    (Section 3.7 — Experiment Protocol, paragraphe 1)
3. **Baseline** :
    * Avant tout scénario, 3 minutes de signal au repos ont été enregistrées pour chaque participant.

    (Section 3.7 — Experiment Protocol, paragraphe 1)
4. **Scénario d'orientation** :
    * Chaque participant a réalisé un scénario de conduite simple sur autoroute pour s'adapter au simulateur et comprendre les niveaux de complexité.

    (Section 3.7 — Experiment Protocol, paragraphe 1)
5. **Scénarios principaux** :
    * Les participants ont réalisé 9 scénarios de 3 minutes chacun.
    * Entre chaque scénario, les participants disposaient de 2 minutes de repos puis 2 minutes de baseline.

    (Section 3.7 — Experiment Protocol, paragraphe 2)
6. **Labelisation** :
    * Toutes les 10 secondes, un signal sonore invite le participant à annoncer verbalement son score PAAS.
    * Un membre de l'équipe note chaque score.

    (Section 3.6 — Cognitive Load Self-Assessment, paragraphe 1)
7. **Suivi du SAS** :
    * Toutes les minutes, les participants évaluaient leur niveau d'inconfort sur une échelle de 1 à 9.
    * 2 sessions ont été interrompues.

    (Section 3.4 — Simulation Adaptation Syndrome, paragraphe 3 ; Section 3.5 — Participants, paragraphe 1)

#### Q13. Pourquoi les auteurs collectent-ils une baseline avant les scénarios ?

**Réponse :**
*(Section 4.3 — Normalization, paragraphe 1 ; Section 3.7 — Experiment Protocol, paragraphes 1 et 2)*

La baseline sert à normaliser les signaux de chaque sujet individuellement. Pour réduire la variabilité entre sujets, chaque valeur de feature est divisée par sa valeur moyenne de baseline. Une baseline de 2 minutes est aussi collectée avant chaque nouveau scénario.

#### Q14. Pourquoi prévoir un scénario d’orientation avant les scénarios principaux ?

**Réponse :**
*(Section 3.7 — Experiment Protocol, paragraphe 1)*

Le scénario d'orientation sert à ce que le participant s'adapte au simulateur, comprenne ce que signifient les niveaux faible/moyen/élevé, et réduise le SAS.


#### Q15. Pourquoi prévoir des périodes de repos entre les scénarios ?

**Réponse :**
*(Section 3.7 — Experiment Protocol, paragraphe 2)*

Les 2 minutes de repos permettent au participant de se reposer, de réduire le risque de SAS et de revenir à un état de charge cognitive plus faible avant le scénario suivant.

#### Q16. Pourquoi demander un score de charge cognitive toutes les 10 secondes plutôt qu’une seule fois à la fin du scénario ?

**Réponse :**
*(Section 4.1 — Pre-processing, paragraphe 1 ; Abstract)*

Chaque scénario est segmenté en 18 fenêtres de 10 secondes. Avoir un score toutes les 10 secondes permet d'associer un label précis à chaque fenêtre de signal. Des labels fréquents permettent aussi une mesure automatique plus fiable et plus dense.

#### Q18. Pourquoi utiliser plusieurs scénarios de conduite au lieu d'un seul scénario ?

**Réponse :**
*(Section 3.3 — Driving Scenarios, paragraphe 1 ; Abstract)*

Chaque scénario a un niveau de complexité désigné. Utiliser 9 scénarios de complexités différentes permet d'induire différents niveaux de charge cognitive, ce qui est indispensable pour avoir des exemples de classes variées à donner au modèle.


#### Q19. Pourquoi peut-on s’attendre à une charge cognitive plus élevée dans les scénarios 8 et 9 ?

**Réponse :**
*(Section 3.3 — Driving Scenarios, Table 3)*

Les scénarios 8 et 9 ont les niveaux de complexité les plus élevés. Le scénario 8 requiert de suivre des instructions tout en réalisant un demi-tour en 3 points, et le scénario 9 demande de naviguer dans une ruelle très étroite en maintenant la précision tout en tentant d'accélérer. Ce sont donc des tâches qui sollicitent simultanément beaucoup de ressources cognitives.


#### Q20. La complexité du scénario suffit-elle à définir le label de charge cognitive ?

**Réponse :**
*(Section 3.6 — Cognitive Load Self-Assessment, paragraphe 1)*

Non. C'est le score PAAS auto-rapporté par le participant qui sert de label. La complexité du scénario est un facteur inducteur, mais le label réel vient de la perception subjective du participant toutes les 10 secondes — deux personnes dans le même scénario peuvent rapporter des scores différents.

## 9. Échantillonnage et segmentation temporelle

Les signaux EEG sont continus. Pour entraîner un modèle, il faut transformer ce signal continu en exemples individuels.

Dans CL-Drive, les scores PAAS de charge cognitive sont collectés toutes les 10 secondes.

Il est donc naturel de découper les signaux physiologiques en fenêtres de 10 secondes :

$$
X_i = \text{segment de 10 s}
$$

$$
y_i = \text{score PAAS associé}
$$

On obtient ainsi le problème suivant :

Le principe d’association segment-label est le suivant :

| Temps | Segment EEG | Score PAAS |
|---|---|---|
| 0–10 s | segment 1 | score 1 |
| 10–20 s | segment 2 | score 2 |
| 20–30 s | segment 3 | score 3 |

## 10. Formulation du problème d'apprentissage automatique

À partir du dataset CL-Drive, on peut formuler plusieurs problèmes de classification.

### Classification binaire

| Scores PAAS | Classe |
|---|---|
| 1 à 4 | charge faible |
| 5 à 9 | charge élevée |

### Classification ternaire

| Scores PAAS | Classe |
|---|---|
| 1 à 3 | charge faible |
| 4 à 6 | charge moyenne |
| 7 à 9 | charge élevée |

Dans le projet, on commencera par la classification binaire.

#### Q27. Pourquoi commencer par une classification binaire plutôt qu’une classification ternaire ?

**Réponse**

*(Section 5 — Benchmarking Results, paragraphe final)*

La classification ternaire est plus complexe que la binaire.

#### Q28. Dans une classification binaire, pourquoi le score PAAS 5 est-il généralement placé dans la classe « charge élevée » dans ce papier ?

**Réponse**

*(Section 4.5 — Training scheme, paragraphe 2 ; Table 1 — Section 2.1)*

Le découpage retenu est [1–4] = faible et [5–9] = élevée. Le score 5 correspond à "ni faible ni élevée" sur l'échelle PAAS. Le document ne justifie pas explicitement ce choix.

#### Q29. Quel est le lien entre ce papier et un système d'interface adaptative ?

**Réponse :**

*(Abstract ; Section 2.3 — BCI, paragraphe 1)*

L'objectif final est de développer des systèmes d'alarme automatisés pour les véhicules intelligents. Si un modèle estime la charge cognitive en temps réel depuis l'EEG, le véhicule peut réagir automatiquement — c'est le principe d'une interface adaptative basée sur un BCI.

## 9. Visualisation des signaux EEG

Avant tout prétraitement ou apprentissage automatique, il est indispensable de visualiser les signaux EEG.

La visualisation permet de :

- vérifier que les données sont correctement chargées ;
- observer les amplitudes ;
- repérer les valeurs aberrantes ;
- détecter les ruptures ou données manquantes ;
- comparer les canaux ;
- repérer des artefacts évidents ;
- comparer qualitativement des segments de faible et forte charge cognitive.

### Algorithme conceptuel pour visualiser un segment EEG

Pour visualiser un segment EEG de 10 secondes :

1. choisir un sujet ;
2. choisir un scénario ;
3. choisir un segment ;
4. récupérer les 4 canaux EEG ;
5. calculer le nombre d’échantillons : $N = f_s \times T$ ;
6. construire l’axe temporel : $t[n] = n/f_s$ ;
7. tracer les quatre canaux en fonction du temps ;
8. ajouter le score PAAS associé au titre du graphique ;
9. commenter la qualité du signal.

#### Q30. Que faut-il regarder en premier lorsqu’on affiche un signal EEG brut ?

**Réponse**

*(Section 4.1 — Pre-processing, paragraphe EEG ; Figure 5)*

Un signal EEG brut présente des amplitudes de l'ordre de ±100 µV. Il faut donc vérifier en premier l'amplitude générale, les valeurs aberrantes, et les artefacts évidents. Le document mentionne que des artefacts ont été retirés par filtrage Butterworth passe-bande 0.4–75 Hz et un filtre coupe-bande à 60 Hz.

#### Q31. Pourquoi afficher les quatre canaux EEG plutôt qu’un seul ?

**Réponse**

*(Section 3.1 — Sensors, paragraphe EEG ; Figure 2a ; Section 2.3 — BCI, paragraphe 2)*

Les 4 canaux couvrent deux régions différentes : frontale (AF7, AF8) et temporale (TP9, TP10). Les régions fronto-pariétales sont particulièrement engagées lors de variations de charge cognitive. Afficher les 4 canaux permet de comparer l'information spatiale et de détecter si un artefact touche un canal en particulier.


#### Q32. Pourquoi comparer visuellement un segment de faible charge et un segment de forte charge cognitive ?

**Réponse**

*(Figure 5 — Section 3.7 — Experiment Protocol)*

Le document propose justement cette comparaison en Figure 5, qui montre des exemples de signaux EEG, ECG, EDA et Gaze en charge faible et élevée. Cela permet de vérifier que les données contiennent bien de l'information discriminante entre les deux classes avant d'entraîner un modèle.

#### Q33. Pourquoi l'utilisation d'un simulateur est-elle à la fois un avantage et une limite ?

**Réponse :**

(Section 3.2 — Experiment Test-bed, paragraphes 1 et 2 ; Section 3.4 — Simulation Adaptation Syndrome, paragraphes 1 et 3)

Avantages :
Le simulateur permet de contrôler précisément les paramètres de conduite : type de véhicule, conditions environnementales, niveau de complexité de chaque scénario.
Il offre une expérience immersive grâce à trois écrans 55 pouces avec vue à 180°, un système de mouvement jusqu'à 100 Hz, et un son directionnel — ce qui rend la simulation réaliste.

Limites :
Le simulateur peut provoquer le Simulation Adaptation Syndrome (SAS) : le décalage entre les entrées visuelles et le système vestibulaire cause des symptômes allant d'un léger inconfort jusqu'aux nausées et vomissements.
Le SAS a conduit à l'interruption de 2 sessions sur 23, ce qui réduit le nombre de sujets exploitables.

**Réponse**

Le papier présente CL-Drive, un dataset multimodal conçu pour estimer la charge cognitive d'un conducteur à partir de signaux physiologiques. En effet, si la charge cognitive devient trop élevée, les performances du conducteur se dégradent et le risque d'accident augmente.

Pour collecter ces données, 21 participants ont conduit dans un simulateur de véhicule. Ils ont réalisé 9 scénarios de 3 minutes chacun, de plus en plus complexe. Cela a permi d'enregistré, de manière simultanée, quatre modalités physiologiques : EEG (via un casque Muse S, à 4 canaux, et de fréquence d'échantillonnage 256 Hz), ECG (via des capteurs Shimmer, de fréquence d'échantillonnage 512 Hz), EDA (via d'autres capteurs Shimmer, de fréquence d'échantillonnage 128 Hz) et Gaze (via un eye tracker Tobii, d'une fréquence d'échantillonnage de 50 Hz). Toutes les 10 secondes, chaque participant donnait verbalement son score PAAS (1 à 9), constituant une vérité terrain sujective mais alignée temporellement avec les fenêtres de signal.

Les signaux enregistrés sont ensuite utilisés pour entraîner des modèles de classification. L'objectif est de prédire, à partir d'une fenêtre de 10 secondes de signal, si la charge cognitive du conducteur est faible (PAAS 1–4) ou élevée (PAAS 5–9) en binaire, ou de la distinguer en trois niveaux faible/moyenne/élevée en ternaire. Plusieurs modèles sont évalués et les meilleurs résultats atteignent 83,67% de précision en classification binaire avec XGBoost sur les 4 modalités combinées.
