## 1. Contexte général

La conduite automobile est une activité complexe qui mobilise plusieurs ressources cognitives : perception visuelle, attention soutenue, prise de décision, anticipation, contrôle moteur et gestion des imprévus.

Le papier CL-Drive s'intéresse donc à une question centrale :

> Peut-on estimer automatiquement la charge cognitive d'un conducteur à partir de signaux physiologiques et comportementaux mesurés pendant la conduite ?

#### Q1. Pourquoi est-il intéressant d'estimer la charge cognitive d'un conducteur ?

**Réponse :**

Il est intéressant d'estimer la charge cognitive d'un conducteur à des fins de sécurité routière. Le papier stipule qu'une partie des accidents de la route chaque année est causée par une charge cognitive élevée. Estimer la charge cognitive permet également aux véhicules intelligents de la mesurer et d'alarmer en cas de charge cognitive élevée. 

## 2. Définition : charge cognitive

#### Q2. Comment définir la charge cognitive ?

**Réponse :**

La charge cognitive est la quantité d'information que notre mémoire de travail est capable de traîter à un moment donné. Il s'agit de la quantité de ressources cognitives requise pour accomplir une tâche. Il y a 2 catégories de charge cognitive :
- intrinsèque : complexité de la tâche en elle-même ;
- extrinsèque : facteurs environnants non liés à la tâche.

#### Q3. Dans le contexte de la conduite, de quels facteurs la charge cognitive peut-elle dépendre ?

**Réponse :**

La charge cognitive dépend à la fois des distractions extérieures (tâches secondaires) et de la complexité intrinsèque de la situation de conduite.

Dans le papier, la charge cognitive est évaluée à l'aide d'une échelle subjective appelée **PAAS**, allant de 1 à 9.

#### Q5. Pourquoi parle-t-on ici d'une vérité terrain subjective ?

**Réponse :**

On parle ici d'une vérité terrain subjective car elle repose sur ce que chaque conducteur ressent et déclare lui-même, et non sur une mesure objective. Deux conducteurs dans la même situation peuvent donc donner des scores différents.

## 5. Modalités mesurées

Le dataset CL-Drive est **multimodal**. Il contient plusieurs types de signaux enregistrés pendant la même tâche de conduite.

#### Q7. Compléter le tableau suivant.

| Modalité | Signification | Signal mesuré | Rôle possible |
|---|---|---|---|
| EEG | Électroencéphalogramme  | Activité électrique du cerveau | Capturer les variations d'activité cérébrale liées à la charge cognitive |
| ECG | Électrocardiogramme | Activité électrique du coeur | Mesurer la variabilité de la fréquence cardiaque (HRV) |
| EDA | Activité électrodermale | Conductance cutanée | Refléter l'activité du système nerveux sympathique |
| Gaze | Suivi oculaire | Mouvements des yeux (pupille, clignement, saccade, fixation) | Corréler les mouvements oculaires avec la charge cognitive |

Dans le projet, la modalité principale sera l’**EEG**. Les autres modalités serviront éventuellement d’extensions pour discuter de la fusion multimodale.

#### Q8. Pourquoi peut-il être intéressant de combiner EEG, ECG, EDA et Gaze pour estimer la charge cognitive ?

**Réponse :**

Il est intéressant de combiner EEG, ECG, EDA et Gaze car chaque signal reflète un aspect distinct de l'organisme à la charge cognitive (activité cérébrale, cardiaque, cutanée, oculaire). La combinaison permet d'avoir une information plus riche et d'améliorer la précision de la classification.

## 7. Protocole expérimental

#### Q11. Quelle est la différence entre un **protocole expérimental** et un **dataset exploitable en machine learning** ?

**Réponse :**

La différence entre un protocole expérimental et un dataset exploitable en machine learning est que le premier définit comment les données sont collectées, tandis que l'autre est le résultat du traitement de ces données brutes pour pouvoir entraîner et évaluer des modèles.

#### Q12. Donner une description détaillée du protocole expérimental utilisé.

**Réponse :**

1. Placement des capteurs : Les capteurs EEG, ECG, EDA et Gaze sont installés sur le participant, puis connectés via Bluetooth à la station de collecte.
2. Baseline initiale (3 minutes) : Un enregistrement de référence est effectué pour permettre une future normalisation des signaux.
3. Scénario d'orientation : Le participant conduit un scénario simple (autoroute à 80 km/h) pour s'adapter au simulateur et comprendre les niveaux de complexité faible, moyen et élevé.
4. Baseline courte (2 minutes) : Avant chaque scénario principal, une nouvelle baseline est enregistrée pour ramener le participant à un état de repos.
5. Scénario de conduite (3 minutes) : Le participant réalise l'un des 9 scénarios de complexité croissante. Toutes les 10 secondes, un signal sonore l'invite à déclarer verbalement son score de charge cognitive (PAAS), noté par un membre de l'équipe.
6. Période de repos (2 minutes) : Le participant se repose pour réduire la fatigue et revenir à un état cognitif de base.
7. Répétition : Les étapes 4 à 6 sont répétées pour chacun des 9 scénarios.

#### Q13. Pourquoi les auteurs collectent-ils une baseline avant les scénarios ?

**Réponse :**

Les auteurs collectent une baseline avant les scénarios car elle sert de référence à chaque participant, ce qui permet de normaliser les signaux physiologiques enregistrés pendant la conduite. C'est important car les signaux EEG, ECG ou EDA varient d'une personne à l'autre.

#### Q14. Pourquoi prévoir un scénario d’orientation avant les scénarios principaux ?

**Réponse :**

Le scénario d'orientation permet :
- au participant de s'habituer au simulateur ;
- lui donner une référence concrète des niveaux de complexité ;
- limiter les risques de Simulator Adaptation Syndrome (SAS).

#### Q15. Pourquoi prévoir des périodes de repos entre les scénarios ?

### Réponse

Les périodes de repos permettent de :
- laisser le participant récupérer physiquement ;
- réduire les risques de SAS ;
- ramener la charge cognitive à un niveau de base avant le scénario suivant.

Cela garantit que chaque scénario est joué dans des conditions comparables.

#### Q16. Pourquoi demander un score de charge cognitive toutes les 10 secondes plutôt qu’une seule fois à la fin du scénario ?

### Réponse

On demande un score de charge cognitive toutes les 10 secondes car, si on le demande une seule fois à la fin, le participant pourrait oublier ce qu'il a ressenti. La période de 10 secondes permet d'obtenir des labels denses et fréquents, ce qui facilite par la suite l'entraînement des modèles de ML.

## 8. Scénarios de conduite

Le dataset contient 9 scénarios principaux, conçus pour induire différents niveaux de charge cognitive.

#### Q17. Compléter le tableau suivant.

| Scénario | Description synthétique |
|---:|---|
| 1 | Conduite sur autoroute à 80 km/h (de jour) |
| 2 | Conduite de nuit à 80 km/h |
| 3 | Conduite de nuit sur autoroute sous la neige à 80 km/h |
| 4 | Frapper une balle de tennis avec le pneu, maintenir la précision et accélérer |
| 5 | Slalom entre des balises, maintenir la précision et accélérer |
| 6 | Passage dans un couloir étroit entre des balises, maintenir la précision et accélérer |
| 7 | Virages à 90 degrés gauche et droite, maintenir la précision et accélérer |
| 8 | Conduite avec instructions vocales, réaliser un demi-tour en 3 points |
| 9 | Navigation dans une ruelle étroite, maintenir la précision et accélérer |

L'hypothèse générale est que les scénarios les plus complexes induisent une charge cognitive plus élevée.

#### Q18. Pourquoi utiliser plusieurs scénarios de conduite au lieu d'un seul scénario ?

**Réponse :**

Utiliser plusieurs scénarios de complexité croissante permet d'induire différents niveaux de charge cognitive de manière contrôlée. Un seul scénario ne produirait qu'un niveau de charge unique, ce qui rendrait impossible l'entraînement d'un modèle capable de distinguer charge faible, moyenne et élevée.

#### Q19. Pourquoi peut-on s’attendre à une charge cognitive plus élevée dans les scénarios 8 et 9 ?

### Réponse

On peut s'attendre à une charge cognitive plus élevée dans les scénarios 8 et 9 car ce sont les plus complexes de l'expérience. Le scénario 8 exige de suivre des instructions vocales tout en réalisant un demi-tour en 3 points sur une route étroite. Le scénario 9 impose de naviguer dans une ruelle très étroite avec précision et vitesse. Ces deux tâches mobilisent simultanément plusieurs ressources cognitives (attention, contrôle moteur, traitement des instructions), ce qui explique les scores PAAS les plus élevés observés.

#### Q20. La complexité du scénario suffit-elle à définir le label de charge cognitive ?

### Réponse

La complexité du scénario ne suffit pas à définir le label de charge cognitive. Le label est un score PAAS déclaré subjectivement par chaque participant. Deux conducteurs dans le même scénario peuvent ressentir des niveaux de charge très différents selon leur expérience de conduite, leur état de fatigue ou leur personnalité. La complexité du scénario est donc un facteur inducteur de charge cognitive, mais ce n'est pas elle qui définit le label : c'est la perception individuelle du conducteur.

## 9. Échantillonnage et segmentation temporelle

Les signaux EEG sont continus. Pour entraîner un modèle, il faut transformer ce signal continu en exemples individuels.

Dans CL-Drive, les scores PAAS de charge cognitive sont collectés toutes les 10 secondes.

Il est donc naturel de découper les signaux physiologiques en fenêtres de 10 secondes :

$$
X_i = \text{segment de 10 s}
$$

$$
y_i = \text{score PAAS associé}
$$

On obtient ainsi le problème suivant :

Le principe d’association segment-label est le suivant :

| Temps | Segment EEG | Score PAAS |
|---|---|---|
| 0–10 s | segment 1 | score 1 |
| 10–20 s | segment 2 | score 2 |
| 20–30 s | segment 3 | score 3 |

#### Q21. Quelle est la fréquence d’échantillonnage EEG utilisée dans le papier ? 


**Réponse :**

La fréquence d'échantillonnage EEG utilisée dans le papier est de 256 Hz. Cela signifie que 256 échantillons sont enregistrés par seconde et par canal.

#### Q22. Combien d’échantillons contient une fenêtre EEG de 10 secondes ?


**Réponse :**

Une fenêtre EEG de 10 secondes contient 2560 échantillons par canal.

Calcul : 256 échantillons/seconde x 10 secondes = 2560 échantillons.

#### Q23. Combien de valeurs contient une fenêtre EEG de 10 secondes si les 4 canaux sont utilisés ?

**Réponse :**

Une fenêtre EEG de 10 secondes avec les 4 canaux contient 10 240 valeurs.

Calcul : 2560 échantillons par canal x 4 canaux = 10 240 valeurs.

#### Q24. Un scénario dure 3 minutes. Combien de fenêtres de 10 secondes peut-on obtenir par scénario ?

### Réponse

On peut obtenir 18 fenêtres de 10 secondes par scénario.

Calcul : 3 minutes = 180 secondes, divisées par fenêtres de 10 secondes = 18 fenêtres.

#### Q25. Si un sujet réalise 9 scénarios complets, combien de fenêtres obtient-on pour ce sujet ?

### Réponse

Un sujet qui réalise 9 scénarios complets produit 162 fenêtres.

Calcul : 18 fenêtres par scénario x 9 scénarios = 162 fenêtres.

#### Q26. Si 21 sujets réalisent tous 9 scénarios complets, combien de fenêtres obtient-on au total ?

### Réponse

Si 21 sujets réalisent tous 9 scénarios complets, on obtient 3402 fenêtres au total.

Calcul : 162 fenêtres par sujet x 21 sujets = 3402 fenêtres.

## 10. Formulation du problème d'apprentissage automatique

À partir du dataset CL-Drive, on peut formuler plusieurs problèmes de classification.

### Classification binaire

| Scores PAAS | Classe |
|---|---|
| 1 à 4 | charge faible |
| 5 à 9 | charge élevée |

### Classification ternaire

| Scores PAAS | Classe |
|---|---|
| 1 à 3 | charge faible |
| 4 à 6 | charge moyenne |
| 7 à 9 | charge élevée |

Dans le projet, on commencera par la classification binaire.

#### Q27. Pourquoi commencer par une classification binaire plutôt qu’une classification ternaire ?

### Réponse

On commence par une classification binaire plutôt que ternaire car elle est plus simple à apprendre et à évaluer. Avec seulement deux classes (charge faible vs charge élevée), le problème est moins ambigu, les frontières de décision sont plus nettes et les performances des modèles sont plus faciles à interpréter. La classification ternaire introduit une classe intermédiaire (charge moyenne) dont les frontières avec les deux autres classes sont naturellement floues, ce qui rend le problème plus difficile.

#### Q28. Dans une classification binaire, pourquoi le score PAAS 5 est-il généralement placé dans la classe « charge élevée » dans ce papier ?

### Réponse

Le score PAAS 5 est placé dans la classe "charge élevée" car il se situe au-dessus du milieu de l'échelle. En l'absence de certitude sur son appartenance, les auteurs choisissent de le classer du côté de la charge élevée (scores 5 à 9) plutôt que faible (scores 1 à 4). Dans le doute, il vaut mieux considérer que le conducteur a une cognitive élevée plutôt que l'inverse. Il ne faudrait pas rater une situation potentiellement dangereuse.

## 9. Visualisation des signaux EEG

Avant tout prétraitement ou apprentissage automatique, il est indispensable de visualiser les signaux EEG.

La visualisation permet de :

- vérifier que les données sont correctement chargées ;
- observer les amplitudes ;
- repérer les valeurs aberrantes ;
- détecter les ruptures ou données manquantes ;
- comparer les canaux ;
- repérer des artefacts évidents ;
- comparer qualitativement des segments de faible et forte charge cognitive.

### Algorithme conceptuel pour visualiser un segment EEG

Pour visualiser un segment EEG de 10 secondes :

1. choisir un sujet ;
2. choisir un scénario ;
3. choisir un segment ;
4. récupérer les 4 canaux EEG ;
5. calculer le nombre d’échantillons : $N = f_s \times T$ ;
6. construire l’axe temporel : $t[n] = n/f_s$ ;
7. tracer les quatre canaux en fonction du temps ;
8. ajouter le score PAAS associé au titre du graphique ;
9. commenter la qualité du signal.
